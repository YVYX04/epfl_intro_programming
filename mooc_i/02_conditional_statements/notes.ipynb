### 1. Branchements conditionnels

Jusqu'ici tous nos traitements n'étaient pas conditionnés en fonction des valeurs que prenaient les variables de nos programmes. Nous introduisons maintenant les structures de contrôles qui permettent de changer ce comportement. 

#### `if` et `else`

Le code ci-dessous expose la structure simple d'un branchement conditionel `if` - `else`. Le mot `else` est optionnel. 

```java
// input
int age = 17;

// conditional statement
if (age < 18)
{
    System.out.println("You cannot drink alcohol.");
} else {
    System.out.println("You can drink alcohol");
}
System.out.println("End of verification.");
```

#### Branchements imbriqués

Les branchements conditionnels imbriqués sont des structures contrôles imbriquées les unes aux autres. Il est recommendé de les utiliser avec modération car le code devient rapidement illisible!

```java
int age = 22;
boolean hasTicket = true;
boolean isVIP = false;

if (age >= 18) {
    if (hasTicket) {
        if (isVIP) {
            System.out.println("Welcome to the VIP lounge!");
        } else {
            System.out.println("Welcome to the main hall!");
        }
    } else {
        if (age < 25) {
            System.out.println("Buy a discounted youth ticket.");
        } else {
            System.out.println("Buy a regular ticket.");
        }
    }
} else {
    System.out.println("Entry is restricted to adults.");
}
```

### 2. Conditions et opérateurs

Nous distinguous les **opérateurs de comparaison** suivants:

```java
int a=1, b=1, c=2;
if(a == b) System.out.println("a = b");
if(a < c) System.out.println("a < c");
if(a > c) System.out.println("This message will not be printed!");
if(b != c) System.out.println("b is not equal to c.");
if(a >= b) System.out.println("a >= b");
```

Nous avons également les **opérateurs logiques** suivants:

```java
boolean p = true;
boolean q = false;
boolean r = true;
boolean s = false;

// logical or
if(p || q) System.out.println("A");

// logical and
if(p && q) System.out.println("B");

// exclusive or
if(p ^ r) System.out.println("C");
if(p ^ q) System.out.println("D"); 

// negation 
if(!(p && q)) System.out.println("B");
```
