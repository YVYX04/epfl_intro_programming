### 1. La boucle `for`

Afin d'éviter les redondances, nous nous pouvons utiliser la boucle `for` qui s'exécute itération par itération, tant qu'une condition données est remplie:

```java
// number of iteration
int N = 5;

// for loop
for(int i = 0; i < N; ++i)
{
    System.out.print(i + ", ");
}
```

Nous pouvons aussi imbriquer ces boucles:

```java
int n = 3, m = 3;
for(int i = 0; i < n; ++i)
{
    for(int j = 0; j < m; ++j)
    {
        if (i == j)   System.out.print("[O]");
        else System.out.print("[ ]");
    }

    System.out.print("\n");
}
```

Il est important de noter la notion de *scope*:

```java
{
    int i = 0;
} // <- i goes out of scope!
System.out.println(i);
```

Dans ce cas la variable `i` a été crée dans une scope différente que le programme principale. Lorsque nous sortons de la scope où `i` a été déclarée, la variable "disparaît". Un autre exemple intéressant:

```java
int i = 0;
{
    int j = 2;
    i += j;
}
System.out.println(i);
```

Dans ce cas, `j` est une variable *locale*.

### 2. La boucle `do-while`

Pour introduire le comportement de cette boucle, considérons l'exemple suivant:

```java
import java.util.Scanner;
Scanner in = new Scanner(System.in);
int positive_input;
do
{
    System.out.println("Pleaser enter a positive integer: ");
    positive_input = in.nextInt();
} while(positive_input <= 0);
```

Les instructions à l'intérieur de la boucle `do...while` sont toujours exécutées au moins une fois. Si la condition de devient jamais fausse, les instructions dans la boucle sont répétées indéfiniment!

```java
// another example
int k = 3, n = 4;
do
{
    ++k;
}while(k <= n);
System.out.println(k);
```

### La boucle `while`

La boucle `while` se comporte de la même manière que la boucle `do...while` à la différence qu'elle n'est pas nécessairement exécutée une fois car la vérification se situe avant la première itération:

```java
int i = 0;
int j = 1;

do{
    System.out.print("do while: " + j);
}while(j < i);

while(j < i) System.out.println(i);
```
