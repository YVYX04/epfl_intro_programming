## 1. Introduction



Jusqu'à présent nous avons étudié des types de données dits "simples". Aujourd'hui, nous découvrons un type plus complexe, les *tableaux* ("arrays" en anglais).

Pour manipuler un ensemble de données comme un tout, nous pouvons utiliser un tableau (ici nous stockons des variables de type `int`):

|age|
|---|
| 18|
|22|
|19|
|19|

### Exemple introductif

Supposons que nous avons $n$ valeurs réelles $x_i \in \mathbb R$. Pour chacune de ses valeurs, nous voulons calculer sa déviation de la moyenne $\overline x = \frac{1}{n} \sum_{i = 1}^{n} x_i$ de cette manière: $x_i - \overline x$. Si nous devions créer un programme qui effectue cette tâche avec nos outils actuels, nous aurions:

````java
int x1 = 110, x2 = 108, x3 = 92, x4 = 89;
int n = 4;
double mean = (double) (x1 + x2 + x3 + x4) / n;
System.out.println("x1 - mean = " + (x1 - mean));
// (...)
````

Ce qui est évidemment une tâche fastidieuse si nous avons un $n$ large. Les tableaux interviennent donc comme un nouveau type de données utile dans ce genre de contexte. Pour créer un tableau d'entiers, nous observons la syntaxe suivante: 

```java
int n = 3;
int array = new int[n];
```

### Définition

Dans notre exemple ci-dessus, nous voyons donc qu'un tableau est typiquememt une collection de variables homogènes. Nous sommes particulièrement intéressés par les cas de figure suivants:

1. tableau à taille connue, avec taille fixe lors de l'exécution
2. tableau à taille inconne, avec taill fixe lors de l'exécution
3. tableau à taille connue, avec taille inconnue lors de l'exécution
4. tableau à taille inconnue, avec taille inconnue lors de l'exécution

Dans le cadre de cette section, nous allons étudier les tableaux de taille fixe.

### Déclaration

Nous formalisons maintenant comment déclarer un tableau. Un tableau se déclare de la même manière que les types simples que nous avons vus, en écrivant le type
puis le nom de la variable. On déclare le type d’un tableau en rajoutant des crochets au type des éléments
de ce tableau. Par exemple, on déclare un tableau d’entiers de la manière suivante :

```java
int values[];
```
ou

```java
int[] values;
```

Voici un exemple où nous affectons des valeurs au tableau:

````java
int tab1[] = {1, 2, 3};
int[] tab2 = {4, 5, 6};
````

Les types évolués sont stockés via des **références**, c’est-à-dire que les variables de ces types ne stockent
pas directement la valeur, mais une référence vers l’endroit où est stockée la vraie valeur. Nous pouvons également initialiser un tableau de taille fixe sans données. La seconde manière est d’initialiser un tableau sans élément, puis de le remplir ailleurs dans le programme.
Dans ce cas, il faut quand même initialiser le tableau avec une taille fixée, et on utilisera la syntaxe suivante :

````java
int[] tab3 = new int[4];
````

Pour remplir ensuite le tableau, on aura besoin d’accéder à ses éléments. On utilise pour cela **l’indexation**,
c’est-à-dire que l’on accède aux éléments selon leur position dans le tableau. On utilisera la syntaxe `tab[i]`
pour accéder au $i$-ème élément du tableau `tab`.

````java
for(int i = 0; i < 3; ++i)
{
    System.out.println("tab1[" + i + "] = " + tab1[i]);
}
````

Il existe un deuxième type de boucle `for`, qui itère sur un ensemble de valeurs :

````java
for(int v : tab1) System.out.println(v);
````

Nous pouvons également examiner la longueur d'un tableau avec la méthode suivante:

````java
int n1 = tab1.length;
System.out.println("Length of tab1: " + n1);
````

L'exemple suivant témoigne de la manière dont la mémoire est altérée pour stocker un tableau:

````java
int tab4[] = {1, 0};
int tab5[] = {0, 1};

tab4[1] = 2;
tab5 = tab4;
++tab5[0];

boolean same = (tab5[1] == tab4[1]) && (tab5[0] == tab4[0]);

if(same) System.out.println(":)");
else System.out.println(":(");
````

En effet, lorsque nous effectuons:
```java
tab5 = tab4;
```
cela signifique que les deux variables pointent vers la même adresse en mémoire, de ce fait, `tab4` et `tab5` sont simplement deux alias qui désignent le même tableau.
