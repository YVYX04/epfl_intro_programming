## 1. Introduction

les fonctions opèrent sur des données et sont donc des *traitements*. Les données influencent le comportement des fonctions. Les fonctions permettent de mieux structurer notre code ainsi que d'éviter les redondances. Par exemple, si nous voulons calculer la valeur du polynôme $p(x) = 2x^2 + 3x - 1$ pour plusieurs valeurs de $x$, nous écrivons:

```java
// p(x)
int p(int x)
{
    return 2*x*x + 3*x - 1;
}

// use the function p(x)
for(int x = 0; x < 5; ++x) System.out.println("p(" + x + ") = " + p(x));
```

## 2. Passage des arguments

En Java, il est important de noter que les arguments sont passés par valeur et donc une référence d'un type évolué va être copiée. Ainsi, lorsque nous manipulons cette référence, l'objet référencé sera modifié:

```java
ArrayList<Integer> arr = new ArrayList();
for(int i = 1; i <= 5; ++i) arr.add(i);
System.out.println("initial array: " + arr);

// method working on that type
void m(ArrayList<Integer> nums)
{
    int n = nums.size();
    for(int i = 0; i < n; ++i) nums.set(i, 2 * nums.get(i));
    System.out.println("new array: " + nums);
    return;
}

m(arr);
System.out.println(arr)
```
