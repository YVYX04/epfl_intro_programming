### 1. Introduction

La programmation orientée objet (POO) est un *paradigme* de programmation particulier. L'un des objectifs principaux de ce paradigme est d'organiser not programmes autour de quatre notions fondamentales:

1. encapsulation
2. abstraction
3. héritage
4. polymorphisme

Ce paradigme remplace la programmation procédurale que nous utilisions précédemment. Par exemple, le code suivant:

````java
// method to compute surface
static double surface(double largeur, double hauteur) {
    return largeur * hauteur;
}

double largeur = 3.0;
double hauteur = 4.0;
System.out.println(surface(largeur, hauteur));
````

Si nous avons besoin de créer trois rectangles, ce programme va devenir extrêmement répétitif et nous il sera difficile d'attribuer à quel rectangle appartient une certaine donnée: mauvaise **tracabilité**. Dans le paradigme de la programmation orientée objet, deux concepts résolvent ces problèmes:

1. **Encapsulation**: Group in one entity the data (attributs) and the operations on them (méthodes). That entity is a classe, which is a new type. A variable of that type is an instance (an objet).

2. **Abstraction**: Identify the characteristics and mechanisms common to all entities, and write one generic description of them (the class), instead of one description per entity.

Together, these notions also mean **hiding**. An object has two levels:

+ **Interface d’utilisation**: the visible part, a few well-chosen methods. This is what the programmeur utilisateur sees. Think of a car’s pedals and steering wheel.
+ **Détails d’implémentation**: the hidden internals. These belong to the programmeur concepteur. Think of the engine.

The rule from the course: attributes are always implementation details. Only some methods form the interface.
This gives robustness. The concepteur can change the inside, and code that uses only the interface keeps working.


#### Two different interfaces

Let us consider these two different interfaces:

````java
// Version A
class Rectangle1 {
    private double largeur;
    private double hauteur;
    public Rectangle1(double l, double h) { largeur = l; hauteur = h; }
    public double surface() { return largeur * hauteur; }
}

// Version B: different implementation, identical interface
class Rectangle2 {
    private double[] dims = new double[2];   // dims[0] = largeur, dims[1] = hauteur
    public Rectangle2(double l, double h) { dims[0] = l; dims[1] = h; }
    public double surface() { return dims[0] * dims[1]; }
}

// Utilisateur's code: works unchanged with A or B
Rectangle1 r1 = new Rectangle1(3.0, 4.0);
Rectangle2 r2 = new Rectangle2(3.0, 4.0);
System.out.println(r1.surface() + " and " + r2.surface());   // 12.0
````

### 2. Classes, objets, attributs, méthodes


````java
class Rectangle {               // declares a new TYPE
    double hauteur;             // attributs: type nom;
    double largeur;

    double surface() {          // méthode: typeRetour nom(params) { corps }
        return hauteur * largeur;
    }
}
````

For the above class, we consider the following important actions:

|Action |	Syntax|	Remark|
|------|-----------|------|
| Declare a variable|	`Rectangle rect1;`	|`rect1` holds a reference, like an array variable |
| Create an instance|	`rect1 = new Rectangle();`	| or both at once: `Rectangle rect1 = new Rectangle();` |
| Access an attribute	| `rect1.hauteur = 3.0;` |	`instance.attribut` |
| Call a method |	`rect1.surface()` |	`instance.methode(args)` |

**Portée de class**: every method has direct access to the class’s attributes. That’s why `surface()` takes no parameters, unlike the procedural `surface(largeur, hauteur)`. Parameters are only for information that comes from outside the object.

````java
class Rectangle {
    double hauteur;
    double largeur;
    double surface() { return hauteur * largeur; }
}

Rectangle rect1 = new Rectangle();
System.out.println(rect1.hauteur);    // 0.0, the default value
rect1.hauteur = 3.0;
rect1.largeur = 4.0;
System.out.println("surface : " + rect1.surface());   // 12.0
````

### 3. `public` / `private`, accesseurs, manipulateurs, `this`

**Theory**

- `private`: the methods flaged as `private` can only be used from inside the class (e.g. by other methods). Typically, the attributes should be private.

- `public`: the interface methods, can be used everywhere.

To give controlled access to private attributes, a class can provide:

- **accesseurs** (getters) return a value: `public double getRayon() { return r; }`
- **manipulateurs** (setters) modify it, can validate the new value, and return nothing: `public void setRayon(double r) { ... }`

With these principles in mind, let us refactor the `Circle` class.

````java
class Circle
{
    // attributes
    private double[] center = {0.0, 0.0};
    private double r = 1.0;

    // --- methods ---
    // * getters *
    public double getRadius() { return this.r; }
    public double getX() { return this.center[0]; } // return coordinate by coordinate to not expose the reference to the array!
    public double getY() { return this.center[1]; }

    // * setters *
    public void setRadius(double r) {
        if(r < 0) {
            throw new IllegalArgumentException("The radius must be >= 0, got : " + r);
        }
        this.r = r;
    }

    // surface
    public double surface()
    {
        return (Math.PI * r * r);
    }
    
    // expand
    public void expand(double lambda)
    {
        /*
        The user enters a value lambda and the radius of the circle is expanded by the lambda.
        */
        if (lambda < 0) {
            throw new IllegalArgumentException("Wrong input, lambda >= 0, got: " + lambda);
        }
        r *= lambda;
    }
}
````

### 6. Constructors: Introduction

A **constructor** is a special method that is automatically invoked when an object is created. It performs everything needed at the start of the object’s life, in particular initializing its attributes. For instance:

````java
class Rectangle
{
    // member variables
    double h;
    double l;

    /**
     * constructor 1
     * @param: double h (height), double l (length)
     */
    public Rectangle(double h, double l)
    {
        this.h = h;
        this.l = l;
    }

    /**
     * surface(): returns the area of the rectangle
     */
    public double surface()
    {
        return l * h;
    }
}
````

We can then use this class as follows:

````java
Rectangle r = new Rectangle(2.0, 5.0);
System.out.println("Surface: " + r.surface());
````

The constructor differs from an ordinary method in three ways:

1. **Name**: Exactly the class name: `Rectangle`
2. **Return type**: None, not even void
3. **Call**: Never called like a method (`r.Rectangle(...)` is impossible). It runs once through `new`.

Constructors can be overloaded: a class can have several constructors with different parameter lists. `new` picks the one whose parameters match the arguments we pass.

````java
class Rectangle {
    private double hauteur;
    private double largeur;

    public Rectangle(double h, double l) {   // constructor 1
        hauteur = h;
        largeur = l;
    }

    public Rectangle(double cote) {          // constructor 2: a square
        hauteur = cote;
        largeur = cote;
    }

    public double surface() { return hauteur * largeur; }
}

Rectangle r1 = new Rectangle(3.0, 4.0);  // → constructor 1
Rectangle r2 = new Rectangle(5.0);       // → constructor 2
System.out.println(r1.surface() + " " + r2.surface());
Rectangle r3 = new Rectangle();       // → once we have decalred constructors, we cannot fallback to the default one
````

Once a class declares at least one constructor, Java stops providing the implicit no-argument constructor. So `new Rectangle()` no longer compiles unless you write `Rectangle()` yourself.

````java
class Test
{
    double a, b;
    public Test(double a, double b)
    {
        this.a = a;
        this.b = b;
    }

    public void printTest()
    {
        System.out.println("[" + a + ", " + b + "]");
    }
}

Test t1 = new Test(1, 1);
t1.printTest();
````

The constructor is chosen at compile time. If no constructor matches, or if two match equally well, it’s a compile error.

### 7. The Default Constructor

**Theory**

First, let us introduce the terminology from the course:

+ **Constructeur par défaut**: any constructor with no parameters, whether we write it or not. Example: `Rectangle() { hauteur = 1.0; largeur = 2.0; }`
+ **Constructeur par défaut par défaut**: the no-parameter constructor that the compiler generates when the class declares no constructor at all. It sets every attribute to its default value: 0 for integers, 0.0 for double, false for boolean, null for objects.

As soon as we declare any constructor, with or without parameters, the generated one is not provided anymore. Declaring constructors is how the designer says "these are the only legal ways to create a Rectangle." If Java kept a hidden `Rectangle()`, it would be a back door around that contract. It would also skip any checks the designer put in the constructor, so other methods could no longer rely on the object’s assumptions (its *invariants*). Because the generated constructor disappears, the user is forced to initialize the object explicitly.

````java
class Rectangle {
    private double hauteur;
    private double largeur;

    public Rectangle()                   { hauteur = 1.0; largeur = 2.0; }    // default constructor (written by you)
    public Rectangle(double c)           { hauteur = c;   largeur = 2.0 * c; }
    public Rectangle(double h, double l) { hauteur = h;   largeur = l; }

    public double getHauteur() { return hauteur; }
}
````

#### calling another constructor with `this(...)`

In Java, a constructor can delegate to another constructor of the same class with `this(...)`. The arguments in `this(...)` select the target constructor, by the same matching rules as for `new`:

```java
public Rectangle(double h, double l) {
    hauteur = h;
    largeur = l;                // the only place that actually assigns
}
public Rectangle() {
    this(0.0, 0.0);             // calls the constructor with two arguments
}
```

There are two rules:

- at most one this(...) call per constructor;
- it must be the very first statement of the constructor.

**default values on attributes**

We can give an attribute a value right where we declare it:

```java
class Rectangle {
    private double hauteur = 0.0;
    private double largeur = 0.0;

    public Rectangle() { }
    public Rectangle(double h, double l) { /* ... */ }
}
```

This value is assigned before the body of any constructor runs. If the constructor called doesn’t modify the attribute, the attribute keeps this value. If the constructor does assign it, the constructor’s value replaces it.

It is still recommended to assign values in the constructors. That way, reading the constructors alone tells us everything the designer intends for initialization.

### 8. The copy constructor

**Theory**

Sometimes you want a new, distinct object that has the same attribute values as an existing one. This is done with a constructor that takes an instance of the same class as its parameter:

```java

    height = otherRect.height;
    length = otherRect.length;

**Conceptual question**: `height` is private. Yet the copy constructor reads `otherRect.height`, which is the private attribute of another object, and it compiles. We first said private members are inaccessible "from outside the class". What does this tell us about what private actually protects? Is it per object or per class?

`private` works per class, not per object. Any code written inside `Rectangle` can access the private members of any `Rectangle` instance, not just `this`.

### 9. End of life and assignment

**Theory**

**End of life**. An object reaches the end of its life when the program no longer holds any reference to it. Example: a `Rectangle` created inside a method and never returned. Its memory should then be freed. In some languages, the programmer must free it explicitly. In Java, the garbage collector (ramasse-miettes) does it automatically: it runs periodically during execution and reclaims unreachable objects.

#### displaying objects with `toString()`

**Theory**

`System.out.println(r1);` doesn’t print the contents of a Rectangle. It prints something like `Rectangle@6d06d69c`: the class name and an identifier linked to the reference. That isn’t useful.

To fix this, Java expects you to write, in the class, a method with exactly this signature:

```java
public String toString() {
    return "Rectangle " + hauteur + " x " + largeur;
}
```

Then `println(r1)` calls it automatically. So does concatenation: `"r1 = " + r1` calls it too.

#### Small exercise

````java
class Duration
{
    // --- member variables ---
    private int hours;
    private int minutes;

    // --- constructors ---
    /**
     * main constructor
     * @param h: number of hours
     * @param m: number of minutes
     */
    public Duration(int h, int m)
    {
        // check positivity
        if(h < 0 || m < 0)
        {
            throw new IllegalArgumentException("Hours and minutes should be positive.");
        }

        // check that minutes are < 60, else increase the hour count
        hours = h + (m / 60);
        minutes = m % 60;
    }

    /** 
     * only minute duration constructor
     * @param totalMinutes, total number of minutes.
    */
    public Duration(int totalMinutes)
    {
        this(0, totalMinutes);
    }

    /**
     * copy constructor
     */
    public Duration(Duration d)
    {
        this(d.hours, d.minutes);
    }

    // --- overriding ---
    public String toString()
    {
        return hours + "h" + ((minutes < 10) ? "0" : "") + minutes;
    }

    // --- overload ---
    public boolean equals(Duration d)
    {
        if(d == null) return false;
        else return ((d.minutes == minutes) && (d.hours == hours));
    }

}

// test
Duration d1 = new Duration(1234);
System.out.println(d1);
````
